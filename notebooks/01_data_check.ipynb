```python
import pandas as pd
from pathlib import Path

proj_root = Path.cwd().parent

df_manuf_2020 = pd.read_excel(proj_root/"data"/"raw"/"Base_de_dados_2020_2021_-_Manufacturing.xlsx")
df_sales_2020 = pd.read_excel(proj_root/"data"/"raw"/"Base_de_dados_2020_2021_-_Sales.xlsx")
df_manuf_2022 = pd.read_excel(proj_root/"data"/"raw"/"Base_de_dados_2022_2023_-_Manufacturing.xlsx")
df_sales_2022 = pd.read_excel(proj_root/"data"/"raw"/"Base_de_dados_2022_2023_-_Sales.xlsx")
```

The goal here is to check whether the data is ready for the OLTP model.

```python
dataframes = {
    "Manufacturing: 2020 - 2021": df_manuf_2020,
    "Sales: 2020 - 2021": df_sales_2020,
    "Manufacturing: 2022 - 2023": df_manuf_2022,
    "Sales 2022 - 2023": df_sales_2022
}

for name, df in dataframes.items():
    print("=" * 30)

    print(f"DATAFRAME == {name}")
    # Checking for null columns
    print(df.isna().any())

    # checking for the values that can be found inside each series and for duplicates
    df_strings = df.select_dtypes(include=["object", "string"])

    for column in df_strings.columns:
        print(f"COLUMN {column} INSIDE {name}")

        print(df_strings[column].unique())
        print(df_strings[column].value_counts())
        print(df_strings[column].duplicated())
```
